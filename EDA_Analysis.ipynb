# Import Required Libraries
This section imports essential libraries for data analysis and visualization, including pandas, matplotlib, and seaborn.

```python
# Import Required Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

# Load and Prepare Data
This section loads the marketing_campaign.xlsx dataset and performs basic data cleaning and preparation.

```python
# Load the dataset
file_path = 'marketing_campaign.xlsx'
df = pd.read_excel(file_path)

# Show shape and first few rows
display(df.shape)
display(df.head())

# Check for missing values
display(df.isnull().sum())
```

```python
# Column types and basic statistics
display(df.dtypes)
display(df.describe(include='all'))
```

```python
# Value counts for categorical columns
for col in df.select_dtypes(include=['object', 'category']).columns:
    display(df[col].value_counts().head(10))
```

## Interpreting the Correlation Matrix
The correlation matrix provides a clear overview of how different numeric factors in our data are related. Here’s what it means for our business decisions:

- If two features have a high positive correlation, it means they tend to increase together. For example, if income and spending are highly correlated, customers with higher income also tend to spend more. This helps us identify key drivers of customer behavior.
- A high negative correlation means as one factor increases, the other decreases. For instance, if age and spending are negatively correlated, older customers may spend less, which can inform targeted marketing strategies.
- Features with values near zero are largely independent, so changes in one do not predict changes in the other.

By understanding these relationships, we can:
- Focus on the most influential factors for our marketing and sales strategies.
- Avoid using redundant information in our models, making our analysis more efficient and actionable.
- Detect potential risks of overlapping information, which could affect the accuracy of our predictions.

In summary, the correlation matrix helps us make smarter, data-driven decisions by revealing which factors matter most and how they interact in our customer base.

```python
# Correlation matrix (numeric columns only)
display(df.select_dtypes(include='number').corr())
```

# Visualize Data with Graphs and Plots
This section creates histograms, heatmaps, and countplots to visualize the dataset.

```python
# Histograms for numeric columns to see the data distribution
df.hist(figsize=(12, 8))
plt.suptitle('Histograms of Numeric Columns')
plt.tight_layout()
plt.show()
```

```python
# Heatmap for correlation
plt.figure(figsize=(10, 6))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Countplot for top categorical column
cat_cols = df.select_dtypes(include=['object', 'category']).columns
if len(cat_cols) > 0:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=df, x=cat_cols[0], order=df[cat_cols[0]].value_counts().index[:10])
    plt.title(f'Countplot of {cat_cols[0]}')
    plt.xticks(rotation=45)
    plt.show()
```

# Analyze Data Using Plots
This section interprets the visualizations to extract insights and perform basic analysis using the plots.

```python
# Outlier detection (IQR method)
for col in df.select_dtypes(include=['number']).columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df[col] < Q1 - 1.5 * IQR) | (df[col] > Q3 + 1.5 * IQR)]
    print(f"{col}: {len(outliers)} outliers")
```

# Additional Custom Visualizations and Insights
This section adds more custom plots and provides explanations for each visualization to help you better understand the data.

## Boxplot for Numeric Columns
Boxplots help visualize the distribution and detect outliers in numeric columns. They show the median, quartiles, and potential outliers.

```python
# Boxplot for all numeric columns
plt.figure(figsize=(12, 6))
sns.boxplot(data=df.select_dtypes(include=['number']))
plt.title('Boxplot for Numeric Columns')
plt.xticks(rotation=45)
plt.show()
```

## Pairplot for Relationships
Pairplots show relationships between pairs of numeric variables and can reveal patterns, clusters, or correlations.

```python
# Pairplot for numeric columns (sample up to 1000 rows for speed)
sns.pairplot(df.select_dtypes(include=['number']).sample(min(1000, len(df))))
plt.suptitle('Pairplot for Numeric Columns', y=1.02)
plt.show()
```

## Distribution Plot for Key Numeric Column
Distribution plots show the shape of the data for a specific numeric column, helping to identify skewness, modality, and spread.

```python
# Distribution plot for the first numeric column
num_cols = df.select_dtypes(include=['number']).columns
if len(num_cols) > 0:
    plt.figure(figsize=(8, 4))
    sns.histplot(df[num_cols[0]], kde=True)
    plt.title(f'Distribution of {num_cols[0]}')
    plt.show()
```

## Bar Plot for Categorical Column vs Numeric
Bar plots show the average value of a numeric column for each category, helping to compare groups and spot trends.

```python
# Bar plot for first categorical vs first numeric column
cat_cols = df.select_dtypes(include=['object', 'category']).columns
num_cols = df.select_dtypes(include=['number']).columns
if len(cat_cols) > 0 and len(num_cols) > 0:
    plt.figure(figsize=(10, 5))
    sns.barplot(x=cat_cols[0], y=num_cols[0], data=df, estimator='mean', ci=None)
    plt.title(f'Average {num_cols[0]} by {cat_cols[0]}')
    plt.xticks(rotation=45)
    plt.show()
```
